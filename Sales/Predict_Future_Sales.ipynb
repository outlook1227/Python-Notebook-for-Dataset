```python
merged_df = pd.merge(items_list_1_df,
                     items_list_2_df,
                     on='item_id',
                     how='outer',
                     indicator=True
                    )
merged_df
```

```python
def process_df(_df):
    # clear nan columns and column "_merge"
    _df = _df.dropna(axis=1).drop(columns='_merge')

    # remove _x and _y in column names
    rename_dict = dict()
    for col_name in _df.columns:
        if '_x' in col_name or '_y' in col_name:
            rename_dict[col_name] = col_name[:-2]

    new_df = _df.rename(columns=rename_dict)
    return new_df
```

```python
items_only_list_1_df = merged_df[merged_df._merge == 'left_only']
items_only_list_1_df = process_df(items_only_list_1_df)
items_only_list_1_df
```

```python
items_only_list_2_df = merged_df[merged_df["_merge"] == 'right_only']
items_only_list_2_df = process_df(items_only_list_2_df)
items_only_list_2_df
```

```python
item_both_list_df = merged_df[merged_df["_merge"] == 'both']
item_both_list_df = process_df(item_both_list_df)
item_both_list_df = item_both_list_df.loc[:,~item_both_list_df.columns.duplicated()]
item_both_list_df
```

```python
print("There are", len(items_only_list_1_df), "items in only list 1")
print("There are", len(items_only_list_2_df), "items in only list 2")
print("There are", len(item_both_list_df), "items in both of the two lists")
```

```python
print("There are", len(items_df), "items in items.csv")
```

```python
# Create check_digit function
def check_digits(name):
    for char in name:
        if char.isdigit():
            return True
    return False
```

```python
items_df['is_digits_in_name'] = items_df["item_name"].apply(check_digits)
items_df
```

```python
item_name_with_digits_df = items_df.loc[items_df['is_digits_in_name']]
item_name_with_digits_df
```

```python
print(f'There are {len(item_name_with_digits_df)} items with digits in item_name')
```

```python
# Create check_fifa function
def check_fifa(name):
    if 'FIFA' in name.upper():
        return True
    return False
```

```python
items_df['is_fifa_in_name'] = items_df['item_name'].apply(check_fifa)
```

```python
item_name_with_fifa_df = items_df[items_df['is_fifa_in_name']]
item_name_with_fifa_df
```

```python
print(f'There are {len(item_name_with_fifa_df)} items with FIFA in item_name')
```
